```python
plt.figure(figsize=(10, 6))  # создаем окно графика размером 10 на 6 дюймов
plt.hist(df["popularity"], bins=20, color="steelblue", edgecolor="black")  # строим гистограмму на 20 столбиков
plt.title("Распределение popularity треков")  # добавляем главный заголовок
plt.xlabel("Popularity")  # подписываем горизонтальную ось х
plt.ylabel("Количество треков")  # подписываем вертикальную ось у
plt.grid(axis="y", alpha=0.3)  # включаем полупрозрачную горизонтальную сетку
plt.show()  # отображаем график на экране
```

Ответы на вопросы


1. Что такое DataFrame?

Это таблица с данными, как в Excel. Только внутри Python.
Строки  это объекты , столбцы  их характеристики  ,просто удобно  работать: фильтровать, сортировать, считать статистику и тыры пыры

---

2. Что такое признак (Feature)

Это один столбец в таблице   то есть одна характеристика объекта
Например, у че то там  песни есть признаки исполнитель, жанр, год выпуска, популярность и так далее  
Каждый столбец = один признак

---

3. Чем среднее отличается от медианы?

- Среднее  складываем все числа и делим на количество
- Медиана  число, которое стоит ровно посередине, если всё отсортировать

Разница вот в чём: среднее сильно ломается от выбросов, а медиана  нет

---
4. Что показывает гистограмма?

Это график столбики, который показывает, как распределены значения
По низу  диапазоны (например, популярность от 0 до 100) по высоте  сколько песен попало в каждый диапазон

---

5. Зачем нужна фильтрация?

Чтобы оставить только нужные строки и выкинуть всё лишнее.

Что бы одной командой  и вместо тысячи строк у тебя только те, что нужны и все емае
